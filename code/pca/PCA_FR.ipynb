```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/content/DMC2_S_CP2.csv")

# 2. Standardize the data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

# 3. Fit PCA
pca = PCA(n_components=0.95)  # retain 95% of variance
data_pca = pca.fit_transform(data_scaled)

# 4. PCA Results
print(f"Original shape: {data.shape}")
print(f"Reduced shape: {data_pca.shape}")

# 5. Explained variance plot
plt.figure(figsize=(10, 6))
plt.plot(np.cumsum(pca.explained_variance_ratio_)*100)
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance (%)")
plt.title("PCA - Explained Variance")
plt.grid(True)
plt.show()

# 6. Optional: Create a DataFrame of reduced data
pca_df = pd.DataFrame(data_pca, columns=[f"PC{i+1}" for i in range(data_pca.shape[1])])
pca_df.to_csv("pca_data_17.csv", index=False)
```

Scree Plot

```python
# Get explained variance values
explained_var = pca.explained_variance_ratio_ * 100
cumulative_var = np.cumsum(explained_var)

# Scree plot (bar + line)
plt.figure(figsize=(10, 6))
plt.bar(range(1, len(explained_var)+1), explained_var, alpha=0.5, align='center', label='Individual Explained Variance')
plt.step(range(1, len(cumulative_var)+1), cumulative_var, where='mid', label='Cumulative Explained Variance', color='blue')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio (%)')
plt.title('Scree Plot with Cumulative Line')
plt.legend(loc='best')
plt.grid(True)
plt.tight_layout()
plt.show()
```
